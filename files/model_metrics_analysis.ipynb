# Model Metrics Analysis
Exploratory analysis of model performance trends and anomaly detection.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load performance metrics
df = pd.read_csv('data/perf-trends.csv')
print(f'Records: {len(df)}')
df.head()
```

## Performance Trend Visualization
Plot AU-ROC and AU-PR over time to identify degradation patterns.

```python
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(df['Month'], df['AU-ROC'], marker='o', label='AU-ROC', color='#60a5fa')
ax.plot(df['Month'], df['AU-PR'], marker='s', label='AU-PR', color='#22d3ee')
ax.axhline(y=0.85, color='#ef4444', linestyle='--', alpha=0.7, label='Threshold')
ax.set_title('Performance Metrics Over Time')
ax.set_ylabel('Score')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Drift Detection
Compute month-over-month drift and flag anomalies.

```python
df['roc_drift'] = df['AU-ROC'].pct_change() * 100
df['drift_flag'] = df['roc_drift'].apply(
    lambda x: 'Drifted' if x < -8 else ('Watch' if x < -5 else 'Healthy')
)

print('\nDrift Analysis:')
print(df[['Month', 'AU-ROC', 'roc_drift', 'drift_flag']].to_string(index=False))
```

```python
# Summary statistics
print(f"Mean AU-ROC: {df['AU-ROC'].mean():.4f}")
print(f"Std AU-ROC:  {df['AU-ROC'].std():.4f}")
print(f"Min AU-ROC:  {df['AU-ROC'].min():.4f}")
print(f"Max AU-ROC:  {df['AU-ROC'].max():.4f}")
print(f"\nDrifted months: {(df['drift_flag'] == 'Drifted').sum()}")
```
